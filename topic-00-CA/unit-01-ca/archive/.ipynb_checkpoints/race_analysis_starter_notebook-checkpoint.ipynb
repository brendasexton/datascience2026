```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

## Load Datasets

```python
athletes = pd.read_csv('athletes.csv')
events = pd.read_csv('events.csv')
races = pd.read_csv('races.csv')
entries = pd.read_csv('race_entries_with_wearables_anomalies.csv')
```

## Inspect Data

```python
athletes.head()
events.head()
races.head()
entries.head()
```

## Merge Datasets for Analysis

```python
df = entries.merge(races, on='raceid', how='left')
df = df.merge(events, on='eventid', how='left')
df = df.merge(athletes, on='athleteid', how='left')
df.head()
```

## Basic Exploratory Analysis

```python
df.describe()
```

## Example Visualisation

```python
sns.histplot(df['avg_heart_rate_numeric'].dropna(), bins=40)
plt.title('Distribution of Average Heart Rate')
plt.show()
```
